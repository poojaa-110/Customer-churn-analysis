```python
df = pd.read_csv("Customer Churn.csv")
print(df['SeniorCitizen'].unique())
```

```python
df['SeniorCitizen'] = df['SeniorCitizen'].replace({
    0: 'No',
    1: 'Yes'
})
```

```python
print(df['SeniorCitizen'].unique())
df.head()
```

conversion of senior citizen value from 0 and 1 to yes or no for easier to understand.

```python
# to see how many customer churn out

plt.figure(figsize=(4,4))
cn=sns.countplot(x='Churn',data=df)
cn.bar_label(cn.containers[0])
plt.title('Churn Counts')
plt.xticks(rotation=75)
plt.tight_layout()
plt.show()
```

```python
cn=df.groupby ('Churn').agg({'Churn':"count"})
plt.figure(figsize=(3,3))
plt.pie(cn["Churn"],labels=cn.index,autopct='%1.2f%%',startangle=45)
plt.title('Percentage of customer Churn' , fontsize=10)
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

from the above visualization we can conclude that 26.54% of customer churned out. 
    lets explore the reason:

```python
plt.figure(figsize=(3,3))
sns.countplot(x='gender', data= df, hue="Churn")
plt.title("Customer churn by Gender")
plt.show()
```

we can conclude from above visualization gender has no effect on customer churn.

```python
plt.figure(figsize=(4,4))
ax=sns.countplot(x = 'SeniorCitizen', data = df)
ax.bar_label(ax.containers[0])
plt.title("Customer churn by Senior Citizen")
plt.show()
```

```python

# Calculate percentage 
data=df.groupby(['SeniorCitizen', 'Churn']).size().groupby(level=0).apply(lambda x: x / x.sum() * 100).unstack()
# Create stacked bar chart
ax = data.plot(
    kind='bar',
    stacked=True,
    figsize=(4,4)
)

# Add percentage labels
for container in ax.containers:
    ax.bar_label(
        container,
        label_type='center',
        fmt='%.1f%%'
    )


plt.title('Customer Churn by Senior Citizen')
plt.xlabel('Senior Citizen')
plt.ylabel('Percentage (%)')
plt.legend(title='Churn')

plt.show()
```

from above chart we can conclude that person under senior citizen category left.

```python
plt.figure(figsize=(12,6))
sns.histplot(x='tenure',data=df,bins=72, hue="Churn")
plt.show()
```

on the basis of tenure we can say that people who use service for long time stayed and people who used service for 1 or 2 month churned out.

```python
plt.figure(figsize=(6,8))
ax=sns.countplot(x='Contract',data=df, hue='Churn')
plt.bar_label(ax.containers[0])
plt.show()
```

from the above chart we can conclude that people who took service for one year contract and two year contract who stayed but people who took
month to month service churn.

```python

columns = [
    'PhoneService',
    'MultipleLines',
    'InternetService',
    'OnlineSecurity',
    'OnlineBackup',
    'DeviceProtection',
    'TechSupport',
    'StreamingTV',
    'StreamingMovies'
]

fig, axes = plt.subplots(3, 3, figsize=(15, 12))

for ax, column in zip(axes.flatten(), columns):
    sns.countplot(
        data=df,
        x=column,
        ax=ax,
        hue="Churn"
    )
    
    ax.set_title(column)
    ax.set_xlabel('')
    ax.set_ylabel('Count')
    
    # Add count labels on bars
    for container in ax.containers:
        ax.bar_label(container)

plt.tight_layout()
plt.show()
```

OnlineSecurity & TechSupport: Customers without these services churn more often. Offering these could be a retention lever.

InternetService: Fiber optic users tend to churn more compared to DSL or no internet service. This may indicate dissatisfaction with fiber service quality or pricing.

OnlineBackup & DeviceProtection: Lack of these services correlates with higher churn, suggesting customers value add-ons that protect their data/devices.

StreamingTV & StreamingMovies: These features don’t show as strong a correlation with churn compared to security/support services. Entertainment add-ons may not be decisive in retention.
PhoneService & MultipleLines: Minimal impact on churn compared to internet-related services.

```python
plt.figure(figsize=(15,10))
ax=sns.countplot(x='PaymentMethod',data=df, hue='Churn')
plt.bar_label(ax.containers[0])
plt.show()
```

Customers paying via electronic check may be less satisfied or more prone to leaving and credit card, bank transfer, and mailed check users show lower churn rates compared to electronic check.These methods are associated with more stable customers.Paperless billing (often linked with electronic check) also correlates with higher churn, reinforcing the idea that this group is more volatile.
